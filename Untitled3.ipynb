```python
import gzip
import re
from datetime import datetime
import pandas as pd

# Helper function to parse log line
def parse_log_line(line):
    # Regex to extract fields from Apache log line
    regex = r'(\S+) (\S+) (\S+) \[([^\]]+)\] "([A-Z]+) ([^"]+) ([^"]+)" (\d{3}) (\d+|-) "([^"]*)" "([^"]*)" (\S+) (\S+)'
    match = re.match(regex, line)
    if match:
        return {
            'ip': match.group(1),
            'time': match.group(4),
            'method': match.group(5),
            'url': match.group(6),
            'status': int(match.group(8)),
            'size': int(match.group(9)) if match.group(9) != '-' else 0,
            'user_agent': match.group(11),
            'vhost': match.group(12),
            'server': match.group(13)
        }
    return None

# Load the GZipped Apache log
def load_log_file(filepath):
    with gzip.open(filepath, 'rt') as f:
        logs = []
        for line in f:
            parsed = parse_log_line(line)
            if parsed:
                logs.append(parsed)
        return pd.DataFrame(logs)

# Convert Apache time format to datetime
def convert_apache_time(time_str):
    return datetime.strptime(time_str, "%d/%b/%Y:%H:%M:%S %z")
```

```python
#1 Load the log data      #Q1
logs = load_log_file('s-anand.net-May-2024.gz')

# Convert time to datetime for filtering
logs['datetime'] = logs['time'].apply(convert_apache_time)
logs['hour'] = logs['datetime'].dt.hour
logs['weekday'] = logs['datetime'].dt.weekday

# Filter for GET requests, status 200-299, under /carnatic/ on Saturdays (weekday == 5)
filtered_logs = logs[
    (logs['method'] == 'GET') &
    (logs['url'].str.startswith('/carnatic/')) &
    (logs['status'] >= 200) & (logs['status'] < 300) &
    (logs['hour'] >= 17) & (logs['hour'] < 21) &
    (logs['weekday'] == 5)
]

# Count the number of successful GET requests
num_requests = len(filtered_logs)
print(f"Successful GET requests: {num_requests}")
```

```python
import json
import pandas as pd

# Load the JSON file
with open('city-product-sales.json') as f:
    sales_data = json.load(f)

# Convert to DataFrame
sales_df = pd.DataFrame(sales_data)
```

```python
from fuzzywuzzy import fuzz

def phonetic_clustering(cities):
    clusters = {}
    for city in cities:
        matched = False
        for cluster in clusters.keys():
            if fuzz.ratio(city, cluster) > 80:  # You can adjust the threshold if needed
                clusters[cluster].append(city)
                matched = True
                break
        if not matched:
            clusters[city] = [city]
    return clusters

# Apply phonetic clustering
city_clusters = phonetic_clustering(sales_df['city'].tolist())

# Create a mapping from original to clustered cities
city_mapping = {name: cluster for cluster, names in city_clusters.items() for name in names}

# Map the city names to clustered cities
sales_df['clustered_city'] = sales_df['city'].map(city_mapping)
```

```python
#9 Filter for product "Hats" and sales >= 49
hats_sales = sales_df[(sales_df['product'] == 'Hats') & (sales_df['sales'] >= 49)]

# Group by clustered city and sum sales
hats_sales_grouped = hats_sales.groupby('clustered_city')['sales'].sum()

# Find the city with the highest unit sales
top_city_hats = hats_sales_grouped.idxmax()
top_sales_hats = hats_sales_grouped.max()

print(f"City with highest Hats sales: {top_city_hats}, Sales: {top_sales_hats}")
```

```python
#10 Filter for product "Juice" and sales >= 55
juice_sales = sales_df[(sales_df['product'] == 'Juice') & (sales_df['sales'] >= 55)]

# Group by clustered city and sum sales
juice_sales_grouped = juice_sales.groupby('clustered_city')['sales'].sum()

# Find the city with the highest unit sales
top_city_juice = juice_sales_grouped.idxmax()
top_sales_juice = juice_sales_grouped.max()

print(f"City with highest Juice sales: {top_city_juice}, Sales: {top_sales_juice}")
```
