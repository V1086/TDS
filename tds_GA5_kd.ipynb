```python
import requests
import numpy as np

# Prompt the user for an API key
api_key = input("Enter your API key: ")

# Define the API endpoint and headers
embedding_url = "https://aiproxy.sanand.workers.dev/openai/v1/embeddings"
headers = {
    "Authorization": f"Bearer {api_key}",
    "Content-Type": "application/json"
}

# Function to get embedding for a given word
def get_embedding(word):
    data = {
        "model": "text-embedding-3-small",
        "input": word
    }
    response = requests.post(embedding_url, headers=headers, json=data)
    response.raise_for_status()
    return response.json()['data'][0]['embedding']

# Function to calculate cosine similarity between two embeddings
def cosine_similarity(embedding1, embedding2):
    a = np.array(embedding1)
    b = np.array(embedding2)
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

# Define the word lists
word_lists = {
    "List 1": ["Optimistic", "Important", "Priority", "Industry"],
    "List 2": ["Evolution", "Available", "Strategy", "Fantastic"],
    "List 3": ["Knowledge", "Animation", "Generation", "Encourage"],
    "List 4": ["Establish", "Majority", "Animation", "Knowledge"]
}


# Fetch embedding for "Abundance"
try:
    embedding_Abundance = get_embedding("Abundance")

    # Calculate and display cosine similarity for each word and the average for each list
    list_averages = {}
    for list_name, words in word_lists.items():
        similarities = []
        print(f"\nCalculating similarities for {list_name}:")
        for word in words:
            embedding_word = get_embedding(word)
            similarity = cosine_similarity(embedding_Abundance, embedding_word)
            similarities.append(similarity)
            print(f"Cosine similarity between 'Abundance' and '{word}': {similarity}")

        # Compute and display average similarity for the list
        average_similarity = sum(similarities) / len(similarities)
        list_averages[list_name] = average_similarity
        print(f"Average cosine similarity for {list_name}: {average_similarity}")

    # Find the list with the lowest average cosine similarity
    lowest_similarity_list = min(list_averages, key=list_averages.get)
    print(f"\nThe list with the lowest average cosine similarity to 'Abundance' is: {lowest_similarity_list}")
    print(f"Average cosine similarity: {list_averages[lowest_similarity_list]}")

except requests.exceptions.RequestException as e:
    print("Error fetching embeddings:", e)
except KeyError:
    print("Error processing the embedding response.")
```

```python
import tiktoken

def calculate_input_cost_cents(text, model="gpt-3.5-turbo-0125", cost_per_million_tokens=50):
    # Get the tokenizer for the specific model
    encoding = tiktoken.encoding_for_model(model)

    # Count tokens
    tokens = encoding.encode(text)
    num_tokens = len(tokens)

    # Calculate cost in cents
    cost_per_token = cost_per_million_tokens / 1_000_000  # cents per token
    total_cost = num_tokens * cost_per_token

    return num_tokens, total_cost

# Test input
text = "Die Kinder spielten fröhlich auf dem Spielplatz, ihre Stimmen erfüllten die Luft."

num_tokens, cost = calculate_input_cost_cents(text)
print(f"Number of tokens: {num_tokens}")
print(f"Cost in cents: {cost:.6f}")
```
